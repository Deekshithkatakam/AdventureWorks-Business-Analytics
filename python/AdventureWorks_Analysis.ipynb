Data cleaning and preparation

```python
#cleaning Calender table
calendar["Date"]= pd.to_datetime(calendar["Date"],format = "%d-%m-%Y")
calendar.info()
```

```python
# cleaning customers table
customers["BirthDate"] = pd.to_datetime(customers["BirthDate"],format = "%d-%m-%Y")
```

```python
#lets create copies of each table so that raw data will be untouched and we can use copies for transformations
customers_clean = customers.copy()
products_clean = products.copy()
calendar_clean = calendar.copy()
sales_clean = sales.copy()
territory_clean = territory.copy()
categories_clean = categories.copy()
subcategories_clean = subcategories.copy()
```

```python
#removing duplicates
customers_clean.duplicated().sum()
customers_clean = customers_clean.drop_duplicates()
customers_clean.duplicated().sum()
```

```python
#Next we will clean missing values in Customers.
customers_clean.isnull().sum()
#we must not fill customer key with 0 or unknown so lets inspect first whether it valid or not
customers_clean[customers_clean["CustomerKey"].isnull()]
#it record is tottally null so
customers_clean = customers_clean.dropna(subset = ["CustomerKey"])
customers_clean["CustomerKey"].isnull().sum()
```

```python
customers_clean[~customers_clean["CustomerKey"].str.isnumeric()]["CustomerKey"]
```

```python
customers_clean[~customers_clean["CustomerKey"].str.isnumeric()]
```

```python
customers_clean = customers_clean[customers_clean["CustomerKey"].str.isnumeric()].copy()
```

```python
customers_clean["CustomerKey"] = customers_clean["CustomerKey"].astype("int64")
```

```python
customers_clean["CustomerKey"].dtype
```

```python
#Handle Missing Values in Customer Attribute
customers_clean.isnull().sum()
```

```python
customers_clean["Gender"] = customers_clean["Gender"].fillna("Unknown")
customers_clean["Prefix"] = customers_clean["Prefix"].fillna("Unknown")
```

```python
customers_clean.isnull().sum()
```

```python
#cleaning products table
products_clean.info()
products_clean.isnull().sum()
```

```python
products_clean["ProductColor"] = products_clean["ProductColor"].fillna("Unknown")
```

```python
products_clean.isnull().sum()
```

```python
#data Modelling & merging
```

```python
#lets validate keys befor merging
sales_clean[["ProductKey","CustomerKey","TerritoryKey"]].head()
```

```python
customers_clean[["CustomerKey"]].head()
```

```python
#lets merge customer and sales
sales_customer = sales_clean.merge(customers_clean, on ="CustomerKey",how = "left")
```

```python
sales_customer_product = sales_customer.merge(products_clean,on = "ProductKey",how = "left")
```

```python
sales_customer_product_sub = sales_customer_product.merge(
    subcategories_clean,
    on="ProductSubcategoryKey",
    how="left"
)
```

```python
sales_customer_product_sub.shape
```

```python
sales_full = sales_customer_product_sub.merge(
    categories_clean,
    on="ProductCategoryKey",
    how="left"
)
```

```python
sales_full.isnull().sum()
```

```python
#here the tables have diff key names so left is for sales final right on for teritory
sales_final = sales_full.merge(
    territory_clean,
    left_on="TerritoryKey",
    right_on="SalesTerritoryKey",
    how="left"
)
```

```python
sales_final.isnull().sum()
```

```python
#Exploratory Data Analysis (EDA)
#checking dataset befor EDA
sales_final.shape
```

```python
#checking summary of total orders, total products sold, total customers, total products.
sales_summary = {
    "Total_Orders" : sales_final["OrderNumber"].nunique(),
    "Total_Products_Sold " : sales_final["OrderQuantity"].sum(),
    "Total_Customers" : sales_final["CustomerKey"].nunique(),
    "Total_Products" : sales_final["ProductKey"].nunique()
}
sales_summary
    
```

```python
#time analysis
sales_final["Year"] = sales_final["OrderDate"].dt.year
sales_final["Month"] = sales_final["OrderDate"].dt.month
sales_final["Month_Name"] = sales_final["OrderDate"].dt.month
```

```python
#Sales Quantity by Year , sales quantity by month
year_sales = (sales_final.groupby("Year")["OrderQuantity"].sum().reset_index())
monthly_sales =(sales_final.groupby(["Year","Month"])["OrderQuantity"].sum().reset_index())
year_sales
```

```python
monthly_sales
```

```python
#total products sold by product category
category_sales = (sales_final.groupby("CategoryName")["OrderQuantity"].sum().reset_index())
category_sales
```

```python
#Top 10 Products
top_products = (sales_final.groupby("ProductName")["OrderQuantity"].sum().reset_index().sort_values("OrderQuantity",ascending = False).head(10))
top_products
```

```python
top_customers = (sales_final.groupby(["CustomerKey","FirstName","LastName"])["OrderQuantity"].sum().reset_index().sort_values("OrderQuantity",ascending = False).head(10))
top_customers
```

```python
territory_sales = (sales_final.groupby("Region")["OrderQuantity"].sum().reset_index().sort_values("OrderQuantity",ascending = False))
territory_sales
```

```python
#This question is slightly different because you're analyzing customer buying behavior by occupation
occupation_sales = (sales_final.groupby("Occupation")["OrderQuantity"].sum().reset_index().sort_values("OrderQuantity", ascending = False).head(10))
occupation_sales
```

```python
education_income = (sales_final.groupby("EducationLevel")["AnnualIncome"].mean().reset_index().sort_values("AnnualIncome", ascending = False))
education_income
```

```python
#gender based analysis which gender bought highest no of products
gender_sales = (sales_final.groupby("Gender")["OrderQuantity"].sum().reset_index().sort_values("OrderQuantity", ascending = False))
gender_sales
```

```python
#Best-selling product in each category
best_selling_product = (sales_final.groupby(["CategoryName","ProductName"])["OrderQuantity"].sum().reset_index().sort_values("OrderQuantity",ascending = False))
best_selling_product["Rank"] = (best_selling_product.groupby("CategoryName")["OrderQuantity"].rank(method ="dense",ascending =False))
best_selling_product = best_selling_product[best_selling_product["Rank"] == 1]
best_selling_product
                                
```

```python
#top3 products of each category
top3_products_category = (sales_final.groupby(["CategoryName","ProductName"])["OrderQuantity"].sum().reset_index())
top3_products_category["Rank"] =(top3_products_category.groupby("CategoryName")["OrderQuantity"].rank(method = "dense",ascending = False))
top3_products_category = top3_products_category[top3_products_category["Rank"] <=3]
top3_products_category
```

```python
#Highest Sales Category in Each Region
top_category_region = (sales_final.groupby(["Region","CategoryName"])["OrderQuantity"].sum().reset_index().sort_values("OrderQuantity",ascending =False))
top_category_region = top_category_region.drop_duplicates(subset ="Region")
top_category_region
```

```python
#Most Popular Product Color
color_sales = (sales_final.groupby("ProductColor")["OrderQuantity"].sum().reset_index().sort_values("OrderQuantity", ascending=False))
color_sales
```

```python
sales_final["CustomerName"] = (
    sales_final["Prefix"] + " " +
    sales_final["FirstName"] + " " +
    sales_final["LastName"]
)
```

```python
#Top 5 Customers in Each Region
top_customers_region = (sales_final
    .groupby(["Region","CustomerKey","CustomerName"])["OrderQuantity"]
    .sum()
    .reset_index())
top_customers_region["Rank"] = top_customers_region.groupby("Region")["OrderQuantity"].rank(method = "dense",ascending = False)
top_customers_region =top_customers_region[top_customers_region["Rank"] <=5]
top_customers_region
```

```python
#Time Analysis
#Age Group Purchases the Most?
#pd.cut() is used to divide continuous numerical values into ranges (bins/categories).
sales_final["Age"] = pd.Timestamp.now().year - sales_final["BirthDate"].dt.year 
sales_final["AgeGroup"] =pd.cut(sales_final["Age"],
                                bins = [18,25,35,45,55,100],
                                labels=["18-25", "26-35", "36-45", "46-55", "56+"],
                                include_lowest = True)
agegroup_sales = (
    sales_final
    .groupby("AgeGroup", observed=False)["OrderQuantity"]
    .sum()
    .reset_index()
    .sort_values("OrderQuantity", ascending=False)
)
agegroup_sales                                
```

```python
#quarter sales
sales_final["Quarter"] = sales_final["OrderDate"].dt.quarter
quarter_sales = (sales_final
    .groupby("Quarter")["OrderQuantity"]
    .sum()
    .reset_index()
    .sort_values("Quarter"))
quarter_sales
```

```python
#country sales
country_sales = (
    sales_final
    .groupby("Country")["OrderQuantity"]
    .sum()
    .reset_index()
    .sort_values("OrderQuantity", ascending=False)
)
country_sales
```

```python
#revenue_analysis
sales_final["Revenue"] = (sales_final["OrderQuantity"] *sales_final["ProductPrice"])
```

```python
#Revenue by Year
year_revenue = (
    sales_final
    .groupby("Year")["Revenue"]
    .sum()
    .reset_index()
)
year_revenue
```

```python
#Revenue by Category
category_revenue = (
    sales_final
    .groupby("CategoryName")["Revenue"]
    .sum()
    .reset_index()
    .sort_values("Revenue", ascending=False)
)
category_revenue
```

```python
#Revenue by Region
region_revenue = (
    sales_final
    .groupby("Region")["Revenue"]
    .sum()
    .reset_index()
    .sort_values("Revenue", ascending=False)
)
region_revenue
```

```python
#Top 10 Revenue Customers
top_revenue_customers = (
    sales_final
    .groupby(["CustomerKey", "CustomerName"])["Revenue"]
    .sum()
    .reset_index()
    .sort_values("Revenue", ascending=False)
    .head(10)
)
top_revenue_customers
```

```python
#Best Revenue Product in Each Category
best_revenue_product = (
    sales_final
    .groupby(["CategoryName", "ProductName"])["Revenue"]
    .sum()
    .reset_index()
)
best_revenue_product["Rank"] = (
    best_revenue_product
    .groupby("CategoryName")["Revenue"]
    .rank(method="dense", ascending=False)
)
best_revenue_product = best_revenue_product[
    best_revenue_product["Rank"] == 1
]
best_revenue_product
```
